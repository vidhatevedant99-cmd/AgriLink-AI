```python
numeric_columns = ["Min Price", "Max Price", "Modal Price", "Arrival Quantity"]
```

```python
for col in numeric_columns:
    df[col] = (df[col]
               .str.replace(",","",regex = False)
                .astype(float) )
```

```python
df["Arrival Date"] = pd.to_datetime(
    df["Arrival Date"], 
    format = "%d-%m-%Y"
)
```

```python
market_price = (
    df.groupby("Market")["Modal Price"]
    .mean()
    .sort_values(ascending = False)
)
market_price.head(15)
```

```python
market_price.tail(15)
```

```python
daily_price  = (
    df.groupby("Arrival Date")["Modal Price"]
    .mean()
    .sort_index()
)
```

```python
daily_price.head()
```

```python
daily_price.tail()
```

```python
daily_price.plot(
    figsize = (12, 5),
    title ="Average Onion price across Maharashtra market"
)
```

```python
market_stats = df.groupby("Market")["Modal Price"].agg(
    ['mean', 'std', 'min', 'max']
)
market_stats.sort_values("std", ascending = False).head()
```

```python
market_stats.sort_values("std").head(15)
```

```python
market_stats["cv"] = (
    market_stats["std"] / market_stats["mean"]
 ) *100
market_stats.sort_values("cv", ascending = False).head()
```

```python
market_stats.sort_values("cv").head(15)
```

```python
#Create a clean dataset copy
clean_df = df.copy()
```

```python
#Kepp the important columns for V1
columns_to_keep = [
        "State/UT",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade",
    "Min Price",
    "Max Price",
    "Modal Price",
    "Price Unit",
    "Arrival Quantity",
    "Arrival Unit",
    "Arrival Date"

]
clean_df = clean_df[columns_to_keep]

#Sort by date
clean_df = clean_df.sort_values("Arrival Date").reset_index(drop = True)

#Check the result
clean_df.head()
```
